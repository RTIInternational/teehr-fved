# Load Primary Timeseries

```python
import teehr
from teehr import RemoteReadWriteEvaluation
from teehr.evaluation.spark_session_utils import create_spark_session
from setup_utils import DEV_LOCATION_ID_LIST, TEEHR_API_KEY
from pathlib import Path
```

```python
# Load API key from variables from file
# See readme.md for instructions
from dotenv import load_dotenv, find_dotenv
load_dotenv(find_dotenv(Path(Path().home(), "teehr-api-key")))
```

```python
spark = create_spark_session()
```

```python
spark.sql("USE iceberg.teehr")
```

```python
ev = RemoteReadWriteEvaluation(spark=spark, enable_spark_proxy=True)
```

```python
spark.sql("SELECT count(*) FROM primary_timeseries").show()
```

```python
ev.download.primary_timeseries(
    primary_location_id=DEV_LOCATION_ID_LIST,
    configuration_name="usgs_observations",
    start_date="2025-10-01",
    end_date="2025-11-01",
    load=True
)
```

```python
spark.sql("SELECT count(*) FROM primary_timeseries").show()
```
