# Load CRMMS Domain Data
Add configurations, variables, and units required for CRMMS output data.

**Note:** When available, locations and crosswalks should be added to this notebook.

```python
import logging

import teehr
from teehr.evaluation.spark_session_utils import create_spark_session

logger = logging.getLogger()
```

```python
spark = create_spark_session()
ev = teehr.RemoteReadWriteEvaluation(spark=spark)
```

```python
ev.configurations.add(
    teehr.Configuration(
        name="crmms_esp",
        timeseries_type="secondary",
        description="CRMMS model with Ensemble Streamflow Prediction (ESP) forecasts"
    )
)
```

```python
ev.configurations.to_sdf().show()
```

```python
ev.variables.add(
    [
        teehr.Variable(
            name="poolelevation_monthly_inst",
            long_name="Monthly Instantaneous Pool Elevation"
        ),
        teehr.Variable(
            name="storage_monthly_inst",
            long_name="Monthly Instantaneous Storage"
        ),
        teehr.Variable(
            name="inflow_monthly_mean",
            long_name="Monthly Mean Inflow"
        ),
        teehr.Variable(
            name="outflow_monthly_mean",
            long_name="Monthly Mean Outflow"
        )
    ]
)
```

```python
ev.variables.to_sdf().show(truncate=False)
```

```python
ev.units.add(
    [
        teehr.Unit(
            name="m^3",
            long_name="Cubic Meters"
        ),
        teehr.Unit(
            name="m",
            long_name="Meters"
        )
    ]
)
```

```python
ev.units.to_sdf().show(truncate=False)
```

```python
spark.stop()
```
