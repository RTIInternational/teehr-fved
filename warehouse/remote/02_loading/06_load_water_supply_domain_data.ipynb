# Load Water Supply Domain Data
Add the configurations, variables and units for water supply data (CBRFC, NRCS and CDEC).

Variables use the FVED schema convention names (`volume_{window}_{statistic}`, with the window as start and end months, e.g. `volume_0407_sum` for Apr-Jul). The local CBRFC and NRCS loaders (`warehouse/local/utils`) still write the older names (e.g. `volume_amjj_sum`) and need renaming before they load to remote.

**Note:** The FVED locations, crosswalks and aliases these sources use (`warehouse/local/locations/`) are not loaded on remote yet.

```python
import teehr
from teehr.evaluation.spark_session_utils import create_spark_session
```

```python
spark = create_spark_session()
ev = teehr.RemoteReadWriteEvaluation(spark=spark)
```

## Configurations

```python
ev.configurations.add(
    [
        teehr.Configuration(
            name="cbrfc_unregulated_esp_forecast",
            timeseries_type="secondary",
            description="CBRFC unregulated ESP water supply volume forecasts",
        ),
        teehr.Configuration(
            name="cbrfc_unregulated_official_forecast",
            timeseries_type="secondary",
            description="CBRFC unregulated official water supply volume forecasts",
        ),
        teehr.Configuration(
            name="cbrfc_unregulated_observed",
            timeseries_type="primary",
            description="CBRFC unregulated water-supply volume observations",
        ),
        teehr.Configuration(
            name="nrcs_adjusted_forecast",
            timeseries_type="secondary",
            description="NRCS adjusted water supply volume forecasts",
        ),
        teehr.Configuration(
            name="nrcs_adjusted_observed",
            timeseries_type="primary",
            description="NRCS adjusted water supply volume observations",
        ),
        teehr.Configuration(
            name="cdec_forecast",
            timeseries_type="secondary",
            description="CDEC forecasts",
        ),
        teehr.Configuration(
            name="cdec_observed",
            timeseries_type="primary",
            description="CDEC observations",
        ),
    ]
)
```

```python
ev.configurations.to_sdf().show(truncate=False)
```

## Variables

```python
ev.variables.add(
    [
        teehr.Variable(name=name, long_name=long_name)
        for name, long_name in {
            "volume_monthly_sum": "Monthly summed water supply volume",
            "volume_0105_sum": "Seasonal summed water supply volume (Jan-May)",
            "volume_0106_sum": "Seasonal summed water supply volume (Jan-Jun)",
            "volume_0207_sum": "Seasonal summed water supply volume (Feb-Jul)",
            "volume_0209_sum": "Seasonal summed water supply volume (Feb-Sep)",
            "volume_0307_sum": "Seasonal summed water supply volume (Mar-Jul)",
            "volume_0309_sum": "Seasonal summed water supply volume (Mar-Sep)",
            "volume_0406_sum": "Seasonal summed water supply volume (Apr-Jun)",
            "volume_0407_sum": "Seasonal summed water supply volume (Apr-Jul)",
            "volume_0409_sum": "Seasonal summed water supply volume (Apr-Sep)",
            "volume_0506_sum": "Seasonal summed water supply volume (May-Jun)",
            "volume_0507_sum": "Seasonal summed water supply volume (May-Jul)",
            "volume_0509_sum": "Seasonal summed water supply volume (May-Sep)",
            "volume_0607_sum": "Seasonal summed water supply volume (Jun-Jul)",
            "volume_0609_sum": "Seasonal summed water supply volume (Jun-Sep)",
            "volume_1009_sum": "Seasonal summed water supply volume (Oct-Sep)",
            "volume_0407_cumsum": "Cumulative water supply volume (Apr-Jul) at each month end",
        }.items()
    ]
)
```

```python
ev.variables.to_sdf().show(truncate=False)
```

## Units

```python
ev.units.add(
    teehr.Unit(
        name="m^3",
        long_name="Cubic Meter"
    )
)
```

```python
ev.units.to_sdf().show(truncate=False)
```

```python
spark.stop()
```
