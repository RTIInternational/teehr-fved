```python
from teehr.evaluation.spark_session_utils import create_spark_session
```

```python
spark = create_spark_session()
```

```python
CATALOG = "iceberg"
NAMESPACES = ["teehr", "schema_evolution"]
```

```python
for ns in NAMESPACES:
    print(f"--- {CATALOG}.{ns} ---")
    rows = spark.sql(f"SHOW TABLES IN {CATALOG}.{ns}").collect()
    for r in rows:
        print(" ", r.tableName)
```

```python
DROP_NAMESPACES_TOO = False  # set True to also drop the empty namespace afterward

for ns in NAMESPACES:
    rows = spark.sql(f"SHOW TABLES IN {CATALOG}.{ns}").collect()
    for r in rows:
        full_name = f"{CATALOG}.{ns}.{r.tableName}"
        print(f"Dropping {full_name}")
        spark.sql(f"DROP TABLE IF EXISTS {full_name} PURGE")
    if DROP_NAMESPACES_TOO:
        spark.sql(f"DROP NAMESPACE IF EXISTS {CATALOG}.{ns}")
```

```python
spark.stop()
```
